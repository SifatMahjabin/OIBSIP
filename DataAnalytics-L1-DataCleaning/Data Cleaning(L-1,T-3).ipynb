```python
df1=df.copy()
```

```python
df1["Age"]=df1["Age"].fillna(df1["Age"].median())
```

```python
df1["Fare"] = df1.groupby("Pclass")["Fare"].transform(lambda s: s.fillna(s.median()))
```

```python
df1["HasCabin"] = df1["Cabin"].notna().astype(int)
df1 = df1.drop(columns=["Cabin"])
```

```python
df1["Sex"] = df1["Sex"].str.strip().str.lower()
df1["Embarked"] = df1["Embarked"].str.strip().str.upper()
```

```python
Q1, Q3 = df["Fare"].quantile([0.25, 0.75])
IQR = Q3 - Q1
lower, upper = Q1 - 1.5 * IQR, Q3 + 1.5 * IQR
outliers = df1[(df1["Fare"] < lower) | (df1["Fare"] > upper)]
print(f"\nFare outliers detected (IQR method): {len(outliers)} "
      f"(bounds: {lower:.2f} to {upper:.2f})")
```

```python
df1["Pclass"] = df1["Pclass"].astype("category")
df1["Sex"] = df1["Sex"].astype("category")
df1["Embarked"] = df1["Embarked"].astype("category")
df1["Fare"] = df1["Fare"].round(2)
```
